```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv("cleaned_dataset.csv")
```

```python
df["condition"].value_counts().plot(kind="bar")

plt.title("Product Condition Distribution")
plt.show()
```

```python
df["location"].value_counts().plot(kind="bar")

plt.title("Listings by Location")
plt.show()
```

```python
df["resale_price"].hist()

plt.title("Resale Price Distribution")

plt.xlabel("Price")

plt.show()
```

```python
df.groupby("condition")["resale_price"].mean().plot(kind="bar")

plt.title("Average Resale Price by Condition")

plt.show()
```
